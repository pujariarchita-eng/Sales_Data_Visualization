```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a sample sales dataset
data = {
    "Month": [
        "January", "February", "March", "April",
        "May", "June", "July", "August",
        "September", "October", "November", "December"
    ],
    "Sales": [
        12000, 15000, 14000, 18000,
        22000, 20000, 25000, 28000,
        24000, 30000, 35000, 40000
    ]
}

df = pd.DataFrame(data)

# 2. Display the dataset
print("SALES DATASET")
print(df.to_string(index=False))

# 3. Display basic information
print("\nDataset Shape:", df.shape)
print("\nDataset Columns:", list(df.columns))
print("\nStatistical Summary:")
print(df["Sales"].describe())

# 4. Create a bar chart
plt.figure(figsize=(10, 5))
plt.bar(df["Month"], df["Sales"])
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

# 5. Create a line chart
plt.figure(figsize=(10, 5))
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

# 6. Analyze the results
highest_month = df.loc[df["Sales"].idxmax(), "Month"]
highest_sales = df["Sales"].max()

lowest_month = df.loc[df["Sales"].idxmin(), "Month"]
lowest_sales = df["Sales"].min()

total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()

print("\nOBSERVATIONS")
print("1. Highest sales occurred in", highest_month,
      "with sales of", highest_sales)
print("2. Lowest sales occurred in", lowest_month,
      "with sales of", lowest_sales)
print("3. Total annual sales:", total_sales)
print("4. Average monthly sales:", round(average_sales, 2))
print("5. Sales generally increased toward the end of the year.")
print("\nTask completed successfully!")
```
